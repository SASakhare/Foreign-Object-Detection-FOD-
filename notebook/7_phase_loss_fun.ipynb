```python
import torch

from src.training.losses.box_loss import (
    box_iou,
    iou_loss
)
```

```python
pred = torch.tensor([
    [10, 10, 50, 50],
    [20, 20, 60, 60]
]).float()

target = torch.tensor([
    [10, 10, 50, 50],
    [30, 30, 60, 60]
]).float()
```

```python
iou = box_iou(
    pred,
    target
)

loss = iou_loss(
    pred,
    target
)

print("IoU:", iou)
print("Loss:", loss)
```

# **DFL Loss**

```python
from src.training.losses.dfl_loss import (
    DistributionFocalLoss
)
```

```python
reg_max = 16

dfl = DistributionFocalLoss(
    reg_max=reg_max
)
```

```python
pred = torch.randn(
    5,
    4,
    reg_max
)
```

```python
target = torch.tensor([
    [2.3, 4.2, 8.7, 3.5],
    [1.2, 5.1, 7.4, 2.8],
    [3.1, 2.5, 6.9, 4.4],
    [4.0, 3.0, 5.0, 2.0],
    [1.5, 4.5, 7.5, 3.5]
])
```

```python
loss = dfl(
    pred,
    target
)

print(
    "DFL:",
    loss.item()
)
```
