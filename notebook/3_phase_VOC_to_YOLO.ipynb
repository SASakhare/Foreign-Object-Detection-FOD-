# **VOCO TO YOLO :**

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

sys.path.append(str(PROJECT_ROOT))

from src.dataset_eda import FODEDA
```

```python
DATASET_ROOT = (
    PROJECT_ROOT /
    "FODPascalVOCFormat-V.2.1"/"FODPascalVOCFormat-V.2.1" /
    "VOC2007"
)

BUILT_ROOT = (
    PROJECT_ROOT /
    "dataset" /
    "built"
)

eda = FODEDA(
    dataset_root=DATASET_ROOT,
    built_root=BUILT_ROOT
)
```

```python
annotations_df = eda.load_annotations()
```

```python
annotations_df.head()
```

```python
class_names = sorted(
    annotations_df["class"]
    .dropna()
    .unique()
    .tolist()
)

class_names
```

```python
from src.voc_to_yolo import VOCToYOLO
```

```python
YOLO_ROOT = (
    PROJECT_ROOT /
    "dataset" /
    "yolo"
)
```

```python
converter = VOCToYOLO(
    dataset_root=DATASET_ROOT,
    built_root=BUILT_ROOT,
    output_root=YOLO_ROOT,
    class_names=class_names
)
```

```python
converter.build()
```

```python
for split in ["train", "val", "test"]:

    image_count = len(
        list(
            (YOLO_ROOT / "images" / split)
            .glob("*.jpg")
        )
    )

    label_count = len(
        list(
            (YOLO_ROOT / "labels" / split)
            .glob("*.txt")
        )
    )

    print(
        f"{split:5s} | "
        f"images: {image_count:5d} | "
        f"labels: {label_count:5d}"
    )
```
