# **DATASET Building :**

### ***Understand the raw dataset***

### 1. `VOC2007/`

* Main folder containing the dataset in **Pascal VOC format**.
* Contains images, annotations, and train/test information.
- https://arxiv.org/pdf/2110.03072
- https://github.com/FOD-UNOmaha/FOD-data

### 2. `JPEGImages/`

* Contains the actual **`.jpg` images**.
* Example:

  * `000000.jpg`
  * `000001.jpg`
  * `000002.jpg`
* These are the images on which our model will detect FOD objects.

### 3. `Annotations/`

* Contains one **`.xml` file for each annotated image**.
* Example:

  * `000000.xml`
  * `000001.xml`
* Each XML contains:

  * Image information
  * Object/class name
  * Bounding-box coordinates
* Example bounding box information:

```text
Class: PlasticPart
xmin
ymin
xmax
ymax
```

* This tells the model **what object is present and where it is**.

### 4. `ImageSets/Main/`

* Contains information about **which images belong to which dataset split**.
* Mainly:

  * `trainval.txt`
  * `test.txt`

### 5. `trainval.txt`

* Contains image IDs intended for the **training/validation pool**.
* We will clean this list and divide it into:

  * Training
  * Validation

### 6. `test.txt`

* Contains image IDs intended for the **test set**.
* We should keep this set **unseen during training**.
* It will be used for final evaluation.

### 7. `CategorizationData/`

This contains additional information about the images.

### 8. `FOD_categorization_annotations.csv`

* Contains **metadata about image conditions**.
* Useful for analyzing:

  * Weather
  * Lighting
  * Other categorization information available in the dataset.
* We'll use it later for **condition-wise analysis**.

### 9. `category_information.txt`

* Explains the meaning of the category values in the CSV.
* For example, your dataset defines:

  * Weather: `0 = Dry`, `1 = Wet`
  * Light: `0 = Bright`, `1 = Dim`, `2 = Dark`
